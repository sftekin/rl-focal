```python
import numpy as np
import glob

ds_name_dict = {
    "bbh": "BIG-Bench Hard (BBH)",
    "gpqa": "Graduate-Level Google-Proof QA",
    "gsm8k": "Grade School Math 8K",
    "mmlu_hf": "Measuring Massive Multitask Language Understanding (MMLU)",
    "musr": "Multistep Soft Reasoning (MuSR)"
}

def pad_array(rewards):
    max_len = max(len(a) for a in rewards)

    padded = np.full((len(rewards), max_len), np.nan)

    for i, a in enumerate(rewards):
        padded[i, :len(a)] = a
    return padded

dataset_name = "mmlu_hf"
offset = 0

select_agent_rewards = []
for reward_pth in glob.glob(f"results/checkpoints/{dataset_name}/exp_*/train_select_agent_rewards.npy"):
    reward = np.load(reward_pth)
    select_agent_rewards.append(reward)

ens_agent_rewards = []
for reward_pth in glob.glob(f"results/checkpoints/{dataset_name}/exp_*/train_ens_agent_rewards.npy"):
    reward = np.load(reward_pth)
    ens_agent_rewards.append(reward)


ens_agent_rewards = pad_array(ens_agent_rewards)[:, :15]
select_agent_rewards = pad_array(select_agent_rewards)[:, :15]
```

```python
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8')

plt.rcParams.update({
        'font.size': 14,              # General font size
        'axes.titlesize': 16,         # Title font size
        'axes.labelsize': 16,         # Axis label font size
        'xtick.labelsize': 16,        # X-axis tick font size
        'ytick.labelsize': 16,        # Y-axis tick font size
        'legend.fontsize': 16         # Legend font size
    })


fig, ax = plt.subplots(1, 2, figsize=(12, 4))
colors = plt.cm.RdBu(np.linspace(0, 1, 10))
reward_names = ["Select Agent", "Ensemble Agent"]
for i, padded in enumerate([select_agent_rewards, ens_agent_rewards]):
    mu = np.nanmean(padded, axis=0)
    std = np.nanstd(padded, axis=0)
    x_axis = np.arange(len(mu))

    ax[i].plot(x_axis, mu, lw=2, color=colors[offset+i+1], zorder=3, label=reward_names[i])
    ax[i].fill_between(x_axis, mu-std, mu+std, color=colors[offset+i+0], alpha=0.3, label="±1 std", zorder=3)
    ax[i].grid(which='major', axis='y', linestyle='-.', zorder=0)
    ax[i].set_xlabel("Episode Number")
    ax[i].set_ylabel("Reward")
    ax[i].set_xlim(x_axis.min()-1, x_axis.max()+1)
    ax[i].set_ylim((mu-std).min() - 1, (mu+std).max() + 1)
    ax[i].legend()

plt.suptitle(f"{ds_name_dict[dataset_name]}")
plt.savefig(f"results/figures/rl_train_curve_{dataset_name}.png", bbox_inches="tight", dpi=200)
```

```python
def moving_accuracy(data, window=20):
    return np.convolve(data, np.ones(window)/window, mode="same")

def cum_acc(data):
    cumulative_acc = np.cumsum(data) / np.arange(1, len(data)+1)
    return cumulative_acc

acc_mean = moving_accuracy(test_errors)
acc_cum = cum_acc(test_errors)
supp_cum = cum_acc(supervised_corrects)
supp_mean = moving_accuracy(supervised_corrects, 200)
```

```python

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader


class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(MLP, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim[0]),
            nn.ReLU(),
            nn.Linear(hidden_dim[0], hidden_dim[1]),
            nn.ReLU(),
            nn.Linear(hidden_dim[1], output_dim),
        )
        self.net.apply(self.init_weights)

    def forward(self, x):
        out = self.net(x)
        out = torch.softmax(out, dim=-1)
        return out

    @staticmethod
    def init_weights(m):
        if isinstance(m, nn.Linear):
            torch.nn.init.xavier_uniform(m.weight)
            m.bias.data.fill_(0.01)

```

```python
from data_generator.data_loader import DataCreator
import numpy as np


task_names = ["bbh", "gsm8k", "mmlu_hf", "gpqa", "musr"]
model_names = "all"
all_data_dict = {}
for task_name in task_names:
    datacreator = DataCreator(task_name, model_names=model_names, task_type="lang")
    max_space_size = 0
    all_data_list = []
    all_label_list = []
    for i, (train_data, test_data, num_models, ds_name) in enumerate(datacreator.load()):
        # print(f"Dataset-{i}: {ds_name}, num models: {num_models}")
        space_size = (train_data.shape[1] - 1) // num_models
        all_data = np.concatenate([train_data, test_data], axis=0)
        all_data_list.append(all_data[:, :-1])
        all_label_list.append(all_data[:, -1])
        if space_size > max_space_size:
            max_space_size = space_size
    all_data_dict[task_name] = [all_data_list, all_label_list]
    print(task_name, max_space_size, num_models)
```

```python
num_models_dict =  {
    "bbh": 6, 
    "gsm8k": 8, 
    "mmlu_hf": 14, 
    "gpqa": 6, 
    "musr": 6
}
```

```python
all_data_padded_dict = {}
for key, (data_arr, label) in all_data_dict.items():
    print(key)
    full_data = []
    for data_r in data_arr:
        for i in range(len(data_r)):
            a = np.stack(np.split(data_r[i], num_models_dict[key]))
            space_size = a.shape[1]
            padded = np.pad(a, ((0, 0), (0, 19-space_size)), mode='constant', constant_values=0)
            full_data.append(padded.flatten())
    all_data_padded_dict[key] = np.stack(full_data)
```

```python

max_len = max(val.shape[1] for key, val in all_data_padded_dict.items())
for key, val in all_data_padded_dict.items():
    print(key, val.shape)


train_data = all_data_padded_dict["bbh"][:int(len(all_data_padded_dict["bbh"]) * 0.8)]
train_label = np.concatenate(all_data_dict["bbh"][1])[:int(len(all_data_padded_dict["bbh"]) * 0.8)]
test_data = []
test_label = []
for key, val in all_data_padded_dict.items():
    ds_len = int(len(val) * 0.8)
    
    pad_width = max_len - val.shape[1]
    padded = np.pad(val, ((0, 0), (0, pad_width)), mode='constant')
    
    test_data.append(padded[ds_len:])
    
    label = all_data_dict[key][1]
    test_label.append(np.concatenate(label)[ds_len:])
    
    
test_data = np.vstack(test_data)
test_label = np.concatenate(test_label)
train_data = np.pad(train_data, ((0, 0), (0, max_len - train_data.shape[1])), mode='constant')
```

```python
from torch.utils.data import DataLoader


train_loader = DataLoader(np.concatenate([train_data, train_label[:, None]], axis=1), batch_size=16, shuffle=True)
test_loader = DataLoader(np.concatenate([test_data, test_label[:, None]], axis=1), batch_size=16, shuffle=True)

```

```python
def test_loop(model, data_loader, ret_correct=False, device="cuda"):
    assert device in ["cuda", "cpu"]
    acc_all = []
    logits = []
    labels = []
    for i, batch_data in enumerate(data_loader):
        in_x = batch_data[:, :-1].to(device).float()
        scores = model(in_x)
        label = batch_data[:, -1].numpy()

        scores = scores.detach().cpu().numpy()
        in_x = in_x.detach().cpu().numpy()
        pred = np.argmax(scores, axis=1)
        corrects = np.sum(pred == label)
        acc_all.append(corrects / len(label) * 100)
        if ret_correct:
            logits.append((pred == label).astype(int))
            labels.append(label)

    acc_all = np.asarray(acc_all)
    acc_mean = np.mean(acc_all)

    if ret_correct:
        logits = np.concatenate(logits)
        labels = np.concatenate(labels)
        return acc_mean, logits, labels
    else:
        return acc_mean
```

```python
space_size = 266
model = MLP(space_size, [100, 100], 19)
model = model.to("cuda")
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

best_val_acc, tol = (0, 0)
for epoch in range(200):
    avg_loss = []
    for i, batch_data in enumerate(train_loader):
        in_x = batch_data[:, :-1].to("cuda").float()
        label = batch_data[:, -1].type(torch.long).to("cuda")

        optimizer.zero_grad()
        out = model(in_x)
        loss = loss_fn(out, label)

        loss.backward()
        optimizer.step()
        avg_loss.append(loss.item())

    if epoch % 10 == 0:
        run_loss = np.mean(avg_loss)
        print(f'Epoch {epoch} | Loss {run_loss:.4f}')


model.eval()
acc_mean, corrects, labels = test_loop(model, test_loader, ret_correct=True)
print(f'Novel Acc = {acc_mean:.4f}')
```
